```python
"""Polymorphism means one name,many forms
in one word: Method overriding
"""
class animal:
    def sound(self):
        print("Animals can make sounds")
class dog(animal):
    def sound(self):
        print("Dog sound is Bark")
class cat(animal):
    def sound(self):
        print("Cat sound is meow")
#polymorphism in action
for animal in [dog(),cat()]:
    animal.sound()
```

```python
"""Another example
"""
class shape:
    def area(self):
        print("Shapes have their area")
class Circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self): #method overriding
        print("Area of circle is : ",3.14*self.radius*self.radius)
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(self): #method overriding
        print("Area of Rectangle is : ",self.length*self.width)
        
c=Circle(5)
r=Rectangle(10,5)
c.area()
r.area()
```

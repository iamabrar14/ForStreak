```python
numbers={0,1,2,3,4,5}
print("Initial Set : ",numbers)
numbers.remove(0)
numbers.add(6)
numbers.update([7,8,9,10])
print("Set after multiple operations : ",numbers)
```

```python
s1={1,2,3,4}
s2={3,4,5,6}
print("Union using 1st syntax (s1|s2) : ",s1|s2)
print("Union using 2nd syntax (s1.union(s2)) : ",s1.union(s2))
print("Intersection using 1st syntax (s1&s2) : ",s1&s2)
print("Intersection using 2nd syntax (s1.intersection(s2)) : ",s1.intersection(s2))
print("Difference : ",s1-s2)
```

```python
animals={"Cat","Dog","Lion"}
print("Animal Set : ",animals)
print("Dog"in animals) #Returing true cause dog is in animals set
print("Tiger" not in animals ) #Returing true cause tiger is not in the animals set
```
